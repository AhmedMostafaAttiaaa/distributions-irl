## 7. 2D visuals

PMF/PDF, CDF, and parameter sweeps with several parameter values on one axis.

```python
x = np.arange(0, 11)
fig, ax = plt.subplots()
ax.bar(x, stats.binom(n=10, p=0.5).pmf(x), color=MAIN)
ax.set_title('Template plot')
ax.set_xlabel('k (count)')
ax.set_ylabel('P(X = k)')
plt.show()
```

## 8. 3D visual

Static matplotlib mplot3d version, then an interactive Plotly version.

```python
fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(projection='3d')
xs, ys = np.meshgrid(np.linspace(-2, 2, 30), np.linspace(-2, 2, 30))
ax.plot_surface(xs, ys, np.exp(-(xs**2 + ys**2)), cmap='viridis')
ax.set_title('Template 3D surface')
ax.set_xlabel('x'); ax.set_ylabel('parameter'); ax.set_zlabel('density')
plt.show()
```

```python
fig = go.Figure(go.Surface(x=xs, y=ys, z=np.exp(-(xs**2 + ys**2))))
fig.update_layout(title='Template interactive surface', width=650, height=450)
fig.show()
```
